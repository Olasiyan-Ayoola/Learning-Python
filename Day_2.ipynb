```python
##write two lines of Python to calculate how many 9.6 kWh (200 Ah) batteries a Nigerian bank branch needs to get 48 kWh of nightly backup power.
```

```python
bank_target_kwh = 48.0
```

```python
usable_factor = 0.85  # Real-world 85% usable capacity buffer
```

```python
energy_kwh = 9.6
```

```python
usable_battery_kwh = energy_kwh * usable_factor
```

```python
batteries_needed = bank_target_kwh / usable_battery_kwh
```

```python
print("Battries capabilities needed by bank is", batteries_needed)
```
